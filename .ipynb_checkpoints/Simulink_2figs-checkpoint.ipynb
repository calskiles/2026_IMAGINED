```python
import dash
from dash import dcc, html, dash_table
from dash.dependencies import Input, Output, State
import matlab.engine
import numpy as np
import plotly.graph_objects as go
import pandas as pd

# Start MATLAB Engine
eng = matlab.engine.start_matlab()

# Load Simulink Model
model_path = '/Users/calskiles/Models/Intermodal_simulation_IBonly_Dash.slx'
eng.load_system(model_path, nargout=0)

# Dash App Setup
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Simulink Intermodal Facility Simulation", style={'fontFamily': 'Arial, Helvetica, sans-serif'}),

    html.Label("Enter Average Hostler Refueling (Charging) Time:"),
    dcc.Input(style={'textAlign': 'right', 'font-size': '12px','width': '80px'}, id="input-param-1", type="number", value=90, step=0.1),

    html.Br(),
    
    html.Label("Enter Average Hostler Move Cycle Time:"),
    dcc.Input(style={'textAlign': 'right', 'font-size': '12px','width': '80px'}, id="input-param-2", type="number", value=20, step=0.1),

    html.Br(),

    html.Label("Enter Number of Hostler Re-Charging (Re-Fueling) Stations:"),
    dcc.Input(style={'textAlign': 'right', 'font-size': '12px','width': '80px'}, id="input-param-3", type="number", value=8, step=1),

    html.Br(),
    
    html.Button("Run Simulation", id="run-button", n_clicks=0),

    dcc.Graph(id="sim-graph-1"),
    dcc.Graph(id="sim-graph-2"),

    html.H3("Average Charger Wait (minutes)"),
    dash_table.DataTable(id="summary-table-1", 
                         columns=[{"name": i, "id": i} for i in ["Statistic", "Value"]],
                         style_table={'width': '50%', 'margin': 'auto'}),

    html.H3("Containers in Staging Area"),
    dash_table.DataTable(id="summary-table-2", 
                         columns=[{"name": i, "id": i} for i in ["Statistic", "Value"]],
                         style_table={'width': '50%', 'margin': 'auto'})
])


@app.callback(
    [Output("sim-graph-1", "figure"),
     Output("sim-graph-2", "figure"),
     Output("summary-table-1", "data"),
     Output("summary-table-2", "data")],
    Input("run-button", "n_clicks"),
    [State("input-param-1", "value"),
     State("input-param-2", "value"),
     State("input-param-3", "value")],
    prevent_initial_call=True
)
def run_simulation(n_clicks, input_value_1, input_value_2, input_value_3):
    try:
        model_name = 'Intermodal_simulation_IBonly_Dash'

        simulation_time = 10000  # Example: User-defined run time
        eng.set_param('Intermodal_simulation_IBonly_Dash', 'StopTime', str(simulation_time), nargout=0)# Set simulation run time dynamically
        
        # Set input parameters in the model
        eng.set_param(f"{model_name}/Hostler Charging/meanChargingTime", 'Value', str(input_value_1), nargout=0)
        eng.set_param(f"{model_name}/Hostler Container Move/meanServiceTime", 'Value', str(input_value_2), nargout=0)
        eng.set_param(f"{model_name}/Hostler Charging/Hostler Charging", 'Capacity', str(input_value_3), nargout=0)

        # Run the simulation
        eng.eval("out = sim('Intermodal_simulation_IBonly_Dash')", nargout=0)
        eng.eval("data = struct(out)", nargout=0)
        data = eng.workspace['data']
        #print(data)

        # Access Charger_Wait Data
        charger_wait_data = data['Data']['Charger_Wait']
        time_data_1 = np.array(charger_wait_data['time']).flatten()
        values_1 = np.array(charger_wait_data['signals']['values']).flatten()

        # Access another dataset, e.g., "Some_Other_Signal"
        other_signal_data = data['Data']['Staging_IBonly']
        time_data_2 = np.array(other_signal_data['time']).flatten()
        values_2 = np.array(other_signal_data['signals']['values']).flatten()

        # Create first figure
        fig1 = go.Figure()
        fig1.add_trace(go.Scatter(x=time_data_1, y=values_1, mode="lines", name="Charger Wait Output"))
        fig1.update_layout(title="Average Charger Wait Time", xaxis_title="Time (minutes)", yaxis_title="Wait Time (minutes)")

        # Create second figure
        fig2 = go.Figure()
        fig2.add_trace(go.Scatter(x=time_data_2, y=values_2, mode="lines", name="Staging Spaces Output"))
        fig2.update_layout(title="Containers in Staging Area", xaxis_title="Time (minutes)", yaxis_title="Staging Spaces in Use")

        # Compute Summary Statistics
        summary_stats_1 = pd.DataFrame({
            "Statistic": ["Mean", "Median", "Min", "Max", "Std Dev"],
            "Value": [np.mean(values_1), np.median(values_1), np.min(values_1), np.max(values_1), np.std(values_1)]
        })

        summary_stats_2 = pd.DataFrame({
            "Statistic": ["Mean", "Median", "Min", "Max", "Std Dev"],
            "Value": [np.mean(values_2), np.median(values_2), np.min(values_2), np.max(values_2), np.std(values_2)]
        })

        # Round values to 2 decimal places
        summary_stats_1["Value"] = summary_stats_1["Value"].round(2)
        summary_stats_2["Value"] = summary_stats_2["Value"].round(2)

        return fig1, fig2, summary_stats_1.to_dict("records"), summary_stats_2.to_dict("records")

    except Exception as e:
        print(f"Error during simulation: {e}")
        return go.Figure(), go.Figure()  # Return empty figures in case of error

    finally:
        if eng is not None:
            eng.quit()


if __name__ == "__main__":
    app.run_server(debug=True)
```
