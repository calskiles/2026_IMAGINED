```python
#------------------------------------------------distnace calc
import pandas as pd
from geopy.distance import geodesic

# Read the new_file.csv
data = pd.read_csv('new_file.csv')

# Calculate the distance using Vincenty's formula and convert to miles
def calculate_distance(row):
    origin = (row['maz_origin_lat'], row['maz_origin_long'])
    destination = (row['maz_dest_lat'], row['maz_dest_long'])
    distance_meters = geodesic(origin, destination).meters
    distance_miles = distance_meters * 0.000621371  # Conversion factor from meters to miles
    return distance_miles

# Apply the distance calculation to each row
data['vincenty_distance_miles'] = data.apply(calculate_distance, axis=1)

# Save the updated DataFrame to new_file.csv
data.to_csv('new_file.csv', index=False)
```

```python
#-------------------------------------------------total travel time calc
import pandas as pd

# Read the CSV file
data = pd.read_csv('your_file.csv')

# Get the available transportation modes
available_modes = ['Car', 'PNR', 'Walk', 'TRW']

# Create new columns for total travel times
for mode in available_modes:
    travel_time_cols = [col for col in data.columns if col.startswith(mode) and ('ivtt' in col or 'Time' in col)]
    total_tt_col = f'{mode}_total_tt'
    data[total_tt_col] = data[travel_time_cols].sum(axis=1)

# Save the updated DataFrame to a new CSV file
data.to_csv('new_file.csv', index=False)
```

```python
#----------------------------------------------------------probabilities of choosing each mode
import pandas as pd
import numpy as np

# Read the data from the file
data = pd.read_csv('output_mode_choice_filter4_diaole_pame.csv')

# Extract the necessary variables from the data
UAM_out_of_veh_tt = data['sens_uam_out_tt3_planb'].values
UAM_Cost = data['sens_uam_cost3_planb'].values
UAM_total_tt = data['sens_uam_speed3_3'].values
Walk_WalkTime = data['Walk_WalkTime'].values
TRW_ivtt = data['TRW_ivtt'].values
TRW_Cost = data['TRW_Cost'].values
TRW_WalkTime = data['TRW_WalkTime'].values
TRW_WaitTime = data['TRW_WaitTime'].values
TRW_Transfer = data['TRW_Transfer'].values
PNR_ivtt = data['PNR_ivtt'].values
PNR_Cost = data['PNR_Cost'].values
PNR_WalkTime = data['PNR_WalkTime'].values
PNR_WaitTime_Tran = data['PNR_WaitTime_Tran'].values
PNR_Transfer = data['PNR_Transfer'].values
Car_Cost = data['Car_Cost'].values
Car_ivtt = data['Car_ivtt'].values

# Function to calculate utility for UAM mode
def calculate_utility_uam(UAM_ivtt, UAM_Cost, UAM_out):
    ASC_UAM = 0.664834*0.5 + 0.969867*0.5
    B_CBD_IVTT = -0.007734
    B_CBD_Cost = -0.178076
    B_CBD_Out_of_veh_tt = -0.031374

    return ASC_UAM + B_CBD_IVTT * UAM_ivtt + B_CBD_Cost * UAM_Cost + B_CBD_Out_of_veh_tt * (UAM_out)


# Function to calculate utility for TRW mode
def calculate_utility_trw(TRW_ivtt, TRW_Cost, TRW_WalkTime, TRW_WaitTime, TRW_Transfer):
    ASC_TRW = 0.772217
    B_CBD_IVTT = -0.007734
    B_CBD_Cost = -0.178076
    B_CBD_Out_of_veh_tt = -0.031374
    B_CBD_Transfer = -0.042995

    return ASC_TRW + B_CBD_IVTT * TRW_ivtt + B_CBD_Cost * TRW_Cost + B_CBD_Out_of_veh_tt * (TRW_WalkTime + TRW_WaitTime) + B_CBD_Transfer * TRW_Transfer


# Function to calculate utility for PNR mode
def calculate_utility_pnr(PNR_ivtt, PNR_Cost, PNR_WalkTime, PNR_WaitTime_Tran, PNR_Transfer):
    ASC_PNR = 0.969867
    B_CBD_IVTT = -0.007734
    B_CBD_Cost = -0.178076
    B_CBD_Out_of_veh_tt = -0.031374
    B_CBD_Transfer = -0.042995

    return ASC_PNR + B_CBD_IVTT * PNR_ivtt + B_CBD_Cost * PNR_Cost + B_CBD_Out_of_veh_tt * (PNR_WalkTime + PNR_WaitTime_Tran) + B_CBD_Transfer * PNR_Transfer


# Function to calculate utility for Car mode
def calculate_utility_car(Car_ivtt, Car_Cost):
    ASC_Car = 0.664834
    B_CBD_IVTT = -0.007734
    B_CBD_Cost = -0.178076

    return ASC_Car + B_CBD_IVTT * Car_ivtt + B_CBD_Cost * Car_Cost


# Function to calculate utility for Walk mode
def calculate_utility_walk(Walk_WalkTime):
    ASC_Walk = 0.0
    B_CBD_WalkTime = -0.049730

    return ASC_Walk + B_CBD_WalkTime * Walk_WalkTime


# Calculate the utilities for each mode based on data
V = np.zeros((len(data), 5))
for i in range(len(data)):
    V[i][0] = calculate_utility_uam(UAM_total_tt[i], UAM_Cost[i], UAM_out_of_veh_tt[i])
    V[i][1] = calculate_utility_trw(TRW_ivtt[i], TRW_Cost[i], TRW_WalkTime[i], TRW_WaitTime[i], TRW_Transfer[i])
    V[i][2] = calculate_utility_pnr(PNR_ivtt[i], PNR_Cost[i], PNR_WalkTime[i], PNR_WaitTime_Tran[i], PNR_Transfer[i])
    V[i][3] = calculate_utility_car(Car_ivtt[i], Car_Cost[i])
    V[i][4] = calculate_utility_walk(Walk_WalkTime[i])

# Calculate choice probabilities
exp_V = np.exp(V)
sum_exp_V = np.sum(exp_V, axis=1, keepdims=True)
choice_probs = exp_V / sum_exp_V

# Calculate average choice probabilities
avg_choice_probs = np.mean(choice_probs, axis=0)

# Print the average choice probabilities for each mode
modes = ['UAM', 'TRW', 'PNR', 'Car', 'Walk']
for i, mode in enumerate(modes):
    print(f"Average probability of choosing {mode}: {avg_choice_probs[i]*100}")

summ = 0
for i, mode in enumerate(modes):
    summ += avg_choice_probs[i]
print(summ)

#UAM_cost = c_base + c_per_mile*distance
#UAM_time = (distance/speed)*congestion_factor + out_of_veh_time
#UAM_time = ground_time*0.45 + out_of_veh_time

#sensitivity for out_of_veh_time, cost, aircraft speed
```

```python
import numpy as np
import pandas as pd

# Set random seed for reproducibility
np.random.seed(42)

# Number of routes
num_routes = 15

# Generate route attributes
routes = pd.DataFrame({
    'route_id': range(1, num_routes + 1),
    'cost': np.random.uniform(10, 100, num_routes),  # cost between 10 and 100
    'travel_time': np.random.uniform(30, 120, num_routes),  # travel time between 30 and 120 minutes
    'toll': np.random.choice([0, 1], num_routes),  # 0 for no toll, 1 for toll
    'congestion_prob': np.random.uniform(0, 1, num_routes),  # congestion probability between 0 and 1
    'scenery': np.random.uniform(1, 5, num_routes),  # scenery rating between 1 and 5
    'safety': np.random.uniform(1, 5, num_routes),  # safety rating between 1 and 5
    'reliability': np.random.uniform(1, 5, num_routes),  # reliability rating between 1 and 5
    'comfort': np.random.uniform(1, 5, num_routes),  # comfort rating between 1 and 5
    'environmental_impact': np.random.uniform(1, 5, num_routes),  # environmental impact rating between 1 and 5
    'road_type': np.random.choice(['highway', 'city street', 'rural road'], num_routes),  # road type
    'accident_rate': np.random.uniform(0, 1, num_routes)  # accident rate between 0 and 1
})

# Mapping road types to numerical values
road_type_mapping = {
    'highway': 1,
    'city street': 2,
    'rural road': 3
}
routes['road_type'] = routes['road_type'].map(road_type_mapping)

print("Routes Data:")
print(routes)

# Number of agents in the population
num_agents = 100

# Generate agent attributes
agents = pd.DataFrame({
    'agent_id': range(1, num_agents + 1)
})

print("\nAgents Data:")
print(agents)

# Methodology: Random Utility Model with Random Beta Coefficients
def random_utility_model(agent, routes):
    # Generate random beta coefficients for this agent
    beta = np.random.uniform(-1, 1, len(routes.columns) - 1)  # exclude 'route_id' column
    
    route_attrs = routes[['cost', 'travel_time', 'toll', 'congestion_prob', 'scenery', 'safety', 'reliability', 'comfort', 'environmental_impact', 'road_type', 'accident_rate']]
    
    # Calculate deterministic utility for each route
    deterministic_utilities = route_attrs.values @ beta
    
    # Add random error term to the deterministic utility
    random_error = np.random.normal(0, 1, len(deterministic_utilities))
    utilities = deterministic_utilities + random_error
    
    # Select top 3 routes with the highest resulting utilities
    top_indices = np.argsort(utilities)[-3:][::-1]  # get indices of the top 3 utilities, sorted in descending order
    consideration_set = routes.iloc[top_indices]['route_id'].values
    return consideration_set

# Assign consideration sets using the Random Utility Model with Random Beta Coefficients
agents['random_consideration_set'] = agents.apply(lambda x: random_utility_model(x, routes), axis=1)

print("\nAgents with Random Utility Consideration Sets:")
print(agents[['agent_id', 'random_consideration_set']].head())
```

```python
import numpy as np
import pandas as pd
from scipy.special import softmax

# Set random seed for reproducibility
np.random.seed(42)

# Number of routes
num_routes = 15

# Generate route attributes
routes = pd.DataFrame({
    'route_id': range(1, num_routes + 1),
    'cost': np.random.uniform(10, 100, num_routes),  # cost between 10 and 100
    'travel_time': np.random.uniform(30, 120, num_routes),  # travel time between 30 and 120 minutes
    'toll': np.random.choice([0, 1], num_routes),  # 0 for no toll, 1 for toll
    'congestion_prob': np.random.uniform(0, 1, num_routes),  # congestion probability between 0 and 1
    'scenery': np.random.uniform(1, 5, num_routes),  # scenery rating between 1 and 5
    'safety': np.random.uniform(1, 5, num_routes),  # safety rating between 1 and 5
    'reliability': np.random.uniform(1, 5, num_routes),  # reliability rating between 1 and 5
    'comfort': np.random.uniform(1, 5, num_routes),  # comfort rating between 1 and 5
    'environmental_impact': np.random.uniform(1, 5, num_routes),  # environmental impact rating between 1 and 5
    'road_type': np.random.choice(['highway', 'city street', 'rural road'], num_routes),  # road type
    'accident_rate': np.random.uniform(0, 1, num_routes)  # accident rate between 0 and 1
})

# Mapping road types to numerical values
road_type_mapping = {
    'highway': 1,
    'city street': 2,
    'rural road': 3
}
routes['road_type'] = routes['road_type'].map(road_type_mapping)

print("Routes Data:")
print(routes)

# Number of agents in the population
num_agents = 100

# Generate agent attributes
agents = pd.DataFrame({
    'agent_id': range(1, num_agents + 1)
})

print("\nAgents Data:")
print(agents)

# Define latent classes and their probabilities
num_classes = 3
class_probabilities = [0.3, 0.5, 0.2]  # Probabilities for each class

# Assign agents to latent classes based on probabilities
agents['latent_class'] = np.random.choice(range(num_classes), size=num_agents, p=class_probabilities)

# Define coefficients for each latent class (example coefficients)
class_betas = {
    0: np.random.uniform(-1, 1, len(routes.columns) - 1),
    1: np.random.uniform(-1, 1, len(routes.columns) - 1),
    2: np.random.uniform(-1, 1, len(routes.columns) - 1)
}

# Methodology: Latent Class Model
def latent_class_model(agent, routes, class_betas):
    # Get the beta coefficients for the agent's latent class
    beta = class_betas[agent['latent_class']]
    
    route_attrs = routes[['cost', 'travel_time', 'toll', 'congestion_prob', 'scenery', 'safety', 'reliability', 'comfort', 'environmental_impact', 'road_type', 'accident_rate']]
    
    # Calculate utility for each route
    utilities = route_attrs.values @ beta
    
    # Calculate the probability of each route using softmax
    probabilities = softmax(utilities)
    
    # Randomly select 3 routes based on their probabilities
    consideration_set = np.random.choice(routes['route_id'], size=3, p=probabilities, replace=False)
    return consideration_set

# Assign consideration sets using the Latent Class Model
agents['latent_class_consideration_set'] = agents.apply(lambda x: latent_class_model(x, routes, class_betas), axis=1)

print("\nAgents with Latent Class Consideration Sets:")
print(agents[['agent_id', 'latent_class', 'latent_class_consideration_set']].head())
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Initialize the graph
G = nx.DiGraph()

# Define nodes
origin_nodes = ['O1', 'O2', 'O3']
destination_node = 'D'
intermediate_nodes = ['I1', 'I2', 'I3']
rc = 0.465
tc = 1.254
# Add nodes to the graph
G.add_nodes_from(origin_nodes + [destination_node] + intermediate_nodes)

# Define edges with their modes, distances, and costs
edges = [
    ('O1', 'I1', 'road', 10, tc),  # example costs and distances
    ('O1', 'I2', 'road', 12, tc),
    ('O2', 'I1', 'road', 8, tc),
    ('O2', 'I2', 'road', 10, tc),
    ('O3', 'D', 'road', 20, tc),
    ('I1', 'I2', 'rail', 5, rc),
    ('I1', 'I3', 'rail', 8, rc),
    ('I2', 'I3', 'rail', 6, rc),
    ('I3', 'D', 'rail', 12, rc),
    ('O3', 'I1', 'road', 18, tc),
    ('O2', 'D', 'road', 16, tc)
]

# Add edges to the graph
for u, v, mode, distance, cost in edges:
    G.add_edge(u, v, mode=mode, distance=distance, cost=cost)

# Define positions for the nodes to reflect the desired layout
pos = {
    'O1': (0, 2),
    'O2': (0, -1),
    'O3': (0, 3),
    'I1': (2, 2),
    'I2': (2, 0),
    'I3': (3, 1),
    'D': (5, 1)
}

# Draw the network
plt.figure(figsize=(10, 6))

# Draw road edges with solid lines
road_edges = [(u, v) for u, v, d in G.edges(data=True) if d['mode'] == 'road']
nx.draw_networkx_edges(G, pos, edgelist=road_edges, style='solid')

# Draw rail edges with dashed lines
rail_edges = [(u, v) for u, v, d in G.edges(data=True) if d['mode'] == 'rail']
nx.draw_networkx_edges(G, pos, edgelist=rail_edges, style='dashed')

# Draw nodes and labels
nx.draw_networkx_nodes(G, pos, node_size=700, node_color='skyblue')
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')

# Draw edge labels (costs)
edge_labels = {(u, v): f'{d["cost"]}' for u, v, d in G.edges(data=True)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='red')

plt.title("Network Graph")
plt.show()
```

```python
import gurobipy as gp
from gurobipy import GRB

# Define sets and parameters
nodes = origin_nodes + intermediate_nodes + [destination_node]
arcs = [(u, v) for u, v, d in G.edges(data=True)]
modes = ['road', 'rail']
od_pairs = [('O1', 'D', 10), ('O2', 'D', 20), ('O3', 'D', 30)]  # Example OD pairs with flow
potential_hubs = intermediate_nodes

# Cost dictionary including distance
costs = {(u, v): G[u][v]['cost'] + G[u][v]['distance'] for u, v in arcs}
fixed_costs = {'I1': 100, 'I2': 100000, 'I3': 100}
discount_rate = 0.05
M = 1000

# Create model
m = gp.Model('IntermodalHubLocation')

# Variables
x = m.addVars(arcs, od_pairs, vtype=GRB.CONTINUOUS, name="x")
z = m.addVars(potential_hubs, vtype=GRB.BINARY, name="z")

# Objective function
m.setObjective(
    gp.quicksum(costs[(i, j)] * x[i, j, o, d, f] for i, j in arcs for o, d, f in od_pairs) +
    discount_rate * gp.quicksum(fixed_costs[i] * z[i] for i in potential_hubs),
    GRB.MINIMIZE
)

# Flow conservation constraints
for i in nodes:
    for o, d, f in od_pairs:
        if i == o:
            m.addConstr(
                gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and i != j) -
                gp.quicksum(x[k, i, o, d, f] for k in nodes if (k, i) in arcs and k != i) == f,
                name=f"flow_conservation_{i}_{o}_{d}"
            )
        elif i == d:
            m.addConstr(
                gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and i != j) -
                gp.quicksum(x[k, i, o, d, f] for k in nodes if (k, i) in arcs and k != i) == -f,
                name=f"flow_conservation_{i}_{o}_{d}"
            )
        else:
            m.addConstr(
                gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and i != j) -
                gp.quicksum(x[k, i, o, d, f] for k in nodes if (k, i) in arcs and k != i) == 0,
                name=f"flow_conservation_{i}_{o}_{d}"
            )

# Hub capacity constraints
for i in potential_hubs:
    for o, d, f in od_pairs:
        m.addConstr(
            gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and G[i][j]['mode'] in modes and i != j) <= M * z[i],
            name=f"hub_capacity_{i}_{o}_{d}"
        )

# Non-negativity constraints
m.addConstrs((x[i, j, o, d, f] >= 0 for i, j in arcs for o, d, f in od_pairs), name="non_negativity")

# Optimize the model
m.optimize()

# Check feasibility and print results
if m.status == GRB.INFEASIBLE:
    print("Model is infeasible. Computing Irreducible Inconsistent Subsystem (IIS)...")
    m.computeIIS()
    print("\nThe following constraints are infeasible:")
    for c in m.getConstrs():
        if c.IISConstr:
            print(f"- {c.ConstrName}")
else:
    print('Optimal objective:', m.objVal)
    for v in m.getVars():
        if v.x > 0:
            print(f'{v.varName}: {v.x}')

# Print hub selections
for i in potential_hubs:
    print(f'Hub {i} selected: {z[i].x}')
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
import gurobipy as gp
from gurobipy import GRB

# Initialize the graph
G = nx.DiGraph()

# Define nodes
origin_nodes = ['O1', 'O2', 'O3']
destination_node = 'D'
intermediate_nodes = ['I1', 'I2', 'I3']

# Define positions (coordinates) for the nodes to reflect the desired layout
pos = {
    'O1': (0, 2),
    'O2': (0, -1),
    'O3': (0, 3),
    'I1': (2, 2),
    'I2': (2, 0),
    'I3': (3, 1),
    'D': (5, 1)
}

# Calculate Euclidean distance
def euclidean_distance(coord1, coord2):
    return np.sqrt((coord1[0] - coord2[0]) ** 2 + (coord1[1] - coord2[1]) ** 2)

# Define edges with their modes
edges = [
    ('O1', 'I1', 'road'),
    ('O1', 'I2', 'road'),
    ('O1', 'D', 'road'),
    ('O2', 'I1', 'road'),
    ('O2', 'I2', 'road'),
    ('O3', 'D', 'road'),
    ('I1', 'I2', 'rail'),
    ('I1', 'I3', 'rail'),
    ('I2', 'I3', 'rail'),
    ('I3', 'D', 'rail'),
    ('O3', 'I1', 'road'),
    ('O2', 'D', 'road')
]

# Define cost multipliers
rc = 0.465
tc = 1.254

# Add edges to the graph with dynamically calculated costs
for u, v, mode in edges:
    distance = euclidean_distance(pos[u], pos[v])
    cost = distance * (tc if mode == 'road' else rc)
    G.add_edge(u, v, mode=mode, distance=distance, cost=cost)

# Draw the network
plt.figure(figsize=(10, 6))

# Draw road edges with solid lines
road_edges = [(u, v) for u, v, d in G.edges(data=True) if d['mode'] == 'road']
nx.draw_networkx_edges(G, pos, edgelist=road_edges, style='solid')

# Draw rail edges with dashed lines
rail_edges = [(u, v) for u, v, d in G.edges(data=True) if d['mode'] == 'rail']
nx.draw_networkx_edges(G, pos, edgelist=rail_edges, style='dashed')

# Draw nodes and labels
nx.draw_networkx_nodes(G, pos, node_size=700, node_color='skyblue')
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')

# Draw edge labels (costs)
edge_labels = {(u, v): f'{d["cost"]:.2f}' for u, v, d in G.edges(data=True)}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='red')

plt.title("Network Graph with Dynamic Distance Costs")
plt.show()

# Define sets and parameters for the optimization model
nodes = origin_nodes + intermediate_nodes + [destination_node]
arcs = [(u, v) for u, v, d in G.edges(data=True)]
od_pairs = [('O1', 'D', 10), ('O2', 'D', 20), ('O3', 'D', 30)]  # Example OD pairs with flow
modes = ['road', 'rail']
potential_hubs = intermediate_nodes

# Cost dictionary including dynamically calculated distance
costs = {(u, v): G[u][v]['cost'] for u, v in arcs}
fixed_costs = {'I1': 10000, 'I2': 10000, 'I3': 10000}
discount_rate = 0.05
M = 1000

# Create model
m = gp.Model('IntermodalHubLocation')

# Variables
x = m.addVars(arcs, od_pairs, vtype=GRB.CONTINUOUS, name="x")
z = m.addVars(potential_hubs, vtype=GRB.BINARY, name="z")

# Objective function
m.setObjective(
    gp.quicksum(costs[(i, j)] * x[i, j, o, d, f] for i, j in arcs for o, d, f in od_pairs) +
    discount_rate * gp.quicksum(fixed_costs[i] * z[i] for i in potential_hubs),
    GRB.MINIMIZE
)

# Flow conservation constraints
for i in nodes:
    for o, d, f in od_pairs:
        if i == o:
            m.addConstr(
                gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and i != j) -
                gp.quicksum(x[k, i, o, d, f] for k in nodes if (k, i) in arcs and k != i) == f,
                name=f"flow_conservation_{i}_{o}_{d}"
            )
        elif i == d:
            m.addConstr(
                gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and i != j) -
                gp.quicksum(x[k, i, o, d, f] for k in nodes if (k, i) in arcs and k != i) == -f,
                name=f"flow_conservation_{i}_{o}_{d}"
            )
        else:
            m.addConstr(
                gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and i != j) -
                gp.quicksum(x[k, i, o, d, f] for k in nodes if (k, i) in arcs and k != i) == 0,
                name=f"flow_conservation_{i}_{o}_{d}"
            )

# Hub capacity constraints
for i in potential_hubs:
    for o, d, f in od_pairs:
        m.addConstr(
            gp.quicksum(x[i, j, o, d, f] for j in nodes if (i, j) in arcs and G[i][j]['mode'] in modes and i != j) <= M * z[i],
            name=f"hub_capacity_{i}_{o}_{d}"
        )

# Non-negativity constraints
m.addConstrs((x[i, j, o, d, f] >= 0 for i, j in arcs for o, d, f in od_pairs), name="non_negativity")

# Optimize the model
m.optimize()

# Check feasibility and print results
if m.status == GRB.INFEASIBLE:
    print("Model is infeasible. Computing Irreducible Inconsistent Subsystem (IIS)...")
    m.computeIIS()
    print("\nThe following constraints are infeasible:")
    for c in m.getConstrs():
        if c.IISConstr:
            print(f"- {c.ConstrName}")
else:
    print('Optimal objective:', m.objVal)
    for v in m.getVars():
        if v.x > 0:
            print(f'{v.varName}: {v.x}')

# Print hub selections
for i in potential_hubs:
    print(f'Hub {i} selected: {z[i].x}')
```

```python
import numpy as np
from sklearn.metrics import pairwise_distances_argmin_min

def capacitated_kmeans(X, n_clusters, max_capacity, max_iter=100, tol=1e-4):
    n_samples = X.shape[0]
    # Step 1: Randomly initialize cluster centers
    initial_indices = np.random.choice(n_samples, n_clusters, replace=False)
    centers = X[initial_indices]
    
    for iteration in range(max_iter):
        # Step 2: Assign points to the nearest cluster center with capacity constraint
        labels = -np.ones(n_samples, dtype=int)
        cluster_sizes = np.zeros(n_clusters, dtype=int)
        
        distances = pairwise_distances_argmin_min(X, centers)[1]
        sorted_indices = np.argsort(distances)
        
        for idx in sorted_indices:
            closest_center = np.argmin(np.linalg.norm(X[idx] - centers, axis=1))
            if cluster_sizes[closest_center] < max_capacity:
                labels[idx] = closest_center
                cluster_sizes[closest_center] += 1
            else:
                # If the closest center is full, find the next closest center that is not full
                for center_idx in np.argsort(np.linalg.norm(X[idx] - centers, axis=1)):
                    if cluster_sizes[center_idx] < max_capacity:
                        labels[idx] = center_idx
                        cluster_sizes[center_idx] += 1
                        break
        
        # Check if all points have been assigned to a cluster
        if np.any(labels == -1):
            print("Warning: Not all points could be assigned to clusters due to capacity constraints.")
            break
        
        # Step 3: Recalculate cluster centers
        new_centers = np.array([X[labels == i].mean(axis=0) for i in range(n_clusters)])
        
        # Check for convergence
        if np.linalg.norm(new_centers - centers) < tol:
            break
        
        centers = new_centers
    
    return labels, centers

# Example usage with latitude and longitude points
X = np.array([
    [41.9, -87.7],
    [41.85, -87.65],
    [41.8, -87.6],
    [41.75, -87.55],
    [41.7, -87.5],
    [41.65, -87.45],
    [41.6, -87.4],
    [41.55, -87.35],
    [41.5, -87.3],
    [41.45, -87.25]
])

# import pandas as pd

# # Read demand points from CSV file
# df = pd.read_csv('demand_points.csv')

# # Assuming the CSV file has columns 'latitude' and 'longitude'
# X = df[['latitude', 'longitude']].values

# # Now you can use X in the capacitated_kmeans function


n_clusters = 3
max_capacity = 3

labels, centers = capacitated_kmeans(X, n_clusters, max_capacity)
print("Cluster Labels:", labels)
print("Cluster Centers:", centers)
```

```python
import gurobipy as gp
from gurobipy import GRB
import random

# Example spatial data (latitude and longitude)
demand_points = {
    0: {"lat": 41.9, "lon": -87.7},
    1: {"lat": 41.85, "lon": -87.65},
    2: {"lat": 41.8, "lon": -87.6},
    3: {"lat": 41.75, "lon": -87.55},
    4: {"lat": 41.7, "lon": -87.5},
    5: {"lat": 41.65, "lon": -87.45},
    6: {"lat": 41.6, "lon": -87.4},
    7: {"lat": 41.55, "lon": -87.35},
    8: {"lat": 41.5, "lon": -87.3},
    9: {"lat": 41.45, "lon": -87.25}
}

# Set random seed for reproducibility
random.seed(1)

# Example number of demand points and facilities
D = len(demand_points)  # Number of demand points
F = 3  # Number of facilities

# Example cost data
C = [100, 150, 120]  # Fixed cost of opening facilities
U = 50  # Cost of uncovered demand
U_c = 20  # Additional cost of uncovered demand in congested regions
d = [[1, 2, 3],    # Cost matrix (distance and time)
     [2, 1, 2],
     [3, 2, 1],
     [4, 3, 2],
     [5, 4, 3],
     [1, 2, 3],
     [2, 1, 2],
     [3, 2, 1],
     [4, 3, 2],
     [5, 4, 3]]

# Capacity and FixedFac
Cap = 3  # Capacity of each facility
FixedFac = 2  # Fixed number of facilities to be opened

# Randomly initialize facility locations within the range of demand points
facilities = []
for f in range(F):
    rand_index = random.randint(0, D-1)
    facilities.append(demand_points[rand_index])

# Model
m = gp.Model()

# Decision Variables
y = m.addVars(F, vtype=GRB.BINARY, name="y")
x = m.addVars(D, F, vtype=GRB.BINARY, name="x")
z = m.addVars(D, vtype=GRB.BINARY, name="z")

# Objective Function
m.setObjective(gp.quicksum(C[j] * y[j] for j in range(F)) +
               U * gp.quicksum(1 - gp.quicksum(x[i, j] for j in range(F)) for i in range(D)) +
               U_c * gp.quicksum(z[i] * (1 - gp.quicksum(x[i, j] for j in range(F))) for i in range(D)) +
               gp.quicksum(d[i][j] * x[i, j] for i in range(D) for j in range(F)), GRB.MINIMIZE)

# Constraints
m.addConstrs((gp.quicksum(x[i, j] for j in range(F)) <= 1 for i in range(D)), "DemandServed")
m.addConstrs((gp.quicksum(x[i, j] for i in range(D)) <= Cap * y[j] for j in range(F)), "Capacity")
m.addConstr(gp.quicksum(y[j] for j in range(F)) == FixedFac, "FixedFacilities")
m.addConstrs((x[i, j] <= y[j] for i in range(D) for j in range(F)), "OpenFacility")
m.addConstrs((x[i, j] == 0 for i in range(D) for j in range(F) if i == j), "NoSelfAssignment")

# Solve
m.optimize()

# Results
if m.status == GRB.OPTIMAL:
    print('Optimal solution found:')
    for j in range(F):
        if y[j].X > 0.5:
            print(f"Facility {j} is opened at latitude {facilities[j]['lat']}, longitude {facilities[j]['lon']}")
            for i in range(D):
                if x[i, j].X > 0.5:
                    print(f"  Demand point {i} is served by facility {j}")
else:
    print('No optimal solution found')
```

```python
import gurobipy as gp
from gurobipy import GRB
import random
import math

# Example spatial data (latitude and longitude)
demand_points = {
    0: {"lat": 41.9, "lon": -87.7, "demand": 1},
    1: {"lat": 41.85, "lon": -87.65, "demand": 1},
    2: {"lat": 41.8, "lon": -87.6, "demand": 1},
    3: {"lat": 41.75, "lon": -87.55, "demand": 1},
    4: {"lat": 41.7, "lon": -87.5, "demand": 1},
    5: {"lat": 41.65, "lon": -87.45, "demand": 1},
    6: {"lat": 41.6, "lon": -87.4, "demand": 1},
    7: {"lat": 41.55, "lon": -87.35, "demand": 1},
    8: {"lat": 41.5, "lon": -87.3, "demand": 1},
    9: {"lat": 41.45, "lon": -87.25, "demand": 1}
}

# import pandas as pd

# # Read the demand points from a CSV file
# df = pd.read_csv('demand_points_all_final.csv')

# # Convert the DataFrame to a dictionary
# demand_points = df.to_dict('index')

# # Adjust the dictionary structure
# demand_points = {i: {"lat": row['lat'], "lon": row['lon'], "demand": row['demand']} for i, row in demand_points.items()}


# Calculate total demand
total_demand = sum(point["demand"] for point in demand_points.values())

# Parameters
Cap = 3  # Capacity of each facility
FixedCost = 100  # Fixed cost of opening a facility
U = 50  # Cost of uncovered demand
U_c = 20  # Additional cost of uncovered demand in congested regions
random.seed(1)  # Set random seed for reproducibility

# Cost matrix (distance between demand points and facilities)
def calculate_distance(p1, p2):
    return ((p1["lat"] - p2["lat"]) ** 2 + (p1["lon"] - p2["lon"]) ** 2) ** 0.5

# Function to solve m-FLP
def solve_m_FLP(demand_points, facilities, Cap, FixedCost, U, U_c):
    D = len(demand_points)
    F = len(facilities)
    d = [[calculate_distance(demand_points[i], facilities[j]) for j in range(F)] for i in range(D)]
    
    m = gp.Model()

    # Decision Variables
    y = m.addVars(F, vtype=GRB.BINARY, name="y")
    x = m.addVars(D, F, vtype=GRB.BINARY, name="x")
    z = m.addVars(D, vtype=GRB.BINARY, name="z")

    # Objective Function
    m.setObjective(gp.quicksum(FixedCost * y[j] for j in range(F)) +
                   U * gp.quicksum(1 - gp.quicksum(x[i, j] for j in range(F)) for i in range(D)) +
                   U_c * gp.quicksum(z[i] * (1 - gp.quicksum(x[i, j] for j in range(F))) for i in range(D)) +
                   gp.quicksum(d[i][j] * x[i, j] for i in range(D) for j in range(F)), GRB.MINIMIZE)

    # Constraints
    m.addConstrs((gp.quicksum(x[i, j] for j in range(F)) <= 1 for i in range(D)), "DemandServed")
    m.addConstrs((gp.quicksum(x[i, j] for i in range(D)) <= Cap * y[j] for j in range(F)), "Capacity")
    m.addConstr(gp.quicksum(y[j] for j in range(F)) == F, "FixedFacilities")
    m.addConstrs((x[i, j] <= y[j] for i in range(D) for j in range(F)), "OpenFacility")

    # Solve
    m.optimize()

    # Collect results
    if m.status == GRB.OPTIMAL:
        T = m.objVal
        solution = {"objective": T, "facilities": [(j, facilities[j]["lat"], facilities[j]["lon"]) for j in range(F) if y[j].X > 0.5]}
        return solution
    else:
        return None

# Initialize facilities within the range of demand points
facilities = []
for _ in range(math.ceil(total_demand / Cap)):
    rand_index = random.randint(0, len(demand_points)-1)
    facilities.append(demand_points[rand_index])

# Greedy Heuristic Algorithm
MinFac = math.ceil(total_demand / Cap)
MaxFac = total_demand

m = MinFac
Tbest = None

while m <= MaxFac:
    solution = solve_m_FLP(demand_points, facilities, Cap, FixedCost, U, U_c)
    print('-----------------------------------------')
    if solution is None:
        break
    T = solution["objective"]
    if Tbest is None or T < Tbest["objective"]:
        Tbest = solution
        m += 1
    else:
        break

# Output the best solution
if Tbest is not None:
    print(f"Best objective value: {Tbest['objective']}")
    for facility in Tbest["facilities"]:
        print(f"Facility {facility[0]} at latitude {facility[1]}, longitude {facility[2]}")
else:
    print("No solution found")
```

```python
import osmnx as ox
import networkx as nx

# Define the latitude and longitude for New Orleans and Saint Louis
origin_lat, origin_lon = 29.9511, -90.0715  # New Orleans, LA
destination_lat, destination_lon = 38.6270, -90.1994  # Saint Louis, MO

# Define the network type (e.g., 'all', 'drive', 'bike', 'walk', 'rail')
# For road network
road_graph = ox.graph_from_point((origin_lat, origin_lon), network_type='drive', dist=1000000)  # Adjust distance as needed

# For rail network (using custom query to get rail network)
rail_graph = ox.graph_from_point((origin_lat, origin_lon), custom_filter='["railway"~"rail"]', dist=1000000)  # Adjust distance as needed

# Merge the two graphs
combined_graph = nx.compose(road_graph, rail_graph)

# Find the nearest nodes in the combined graph to the origin and destination points
origin_node = ox.distance.nearest_nodes(combined_graph, X=origin_lon, Y=origin_lat)
destination_node = ox.distance.nearest_nodes(combined_graph, X=destination_lon, Y=destination_lat)

# Get the shortest path between the origin and destination nodes
shortest_path = nx.shortest_path(combined_graph, origin_node, destination_node, weight='length')

# Create a subgraph from the shortest path
subgraph = combined_graph.subgraph(shortest_path).copy()

# Save the subgraph as a NetworkX graph file
nx.write_gpickle(subgraph, 'road_rail_network.gpickle')

print("Network has been saved as road_rail_network.gpickle")
```
